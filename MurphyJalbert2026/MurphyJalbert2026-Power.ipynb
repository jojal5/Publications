### Utilities

```julia
function changerate(x::Vector{<:Real})
    
    n = length(x)
   
    c = count(diff(x) .≈ 0.)
    
    return (n-c)/n
    
end

function changerate(X::Matrix{<:Real})
    
    n = size(X,1)
   
    c = vec(count(diff(X,dims=1) .≈ 0., dims=1))
    
    return (n .- c)./n
    
end
```

### Fitting the spatial model

- κ and σ as a linear function of the covariate
- Common ξ for all stations

```julia
function loglike(y::Vector{<:Real}, n::Int, threshold::Real, η::Real, ϕ::Real, ξ::Real)
    κ = exp(η)
    σ = exp(ϕ)
        
    pd = ExtendedGeneralizedPareto(Power(κ), GeneralizedPareto(σ, ξ))
        
    if n == 0
        return sum(logpdf(pd, y))
    else
        return sum(logpdf(pd, y)) + n*logcdf(pd, threshold)
    end
end
```

```julia
NITER = 20000
m = length(y)

x = (covariate.x .- mean(covariate.x))./std(covariate.x)

X = [ones(length(x)) x]

Q = X'X + I
V = inv(X'X +I)

η = Matrix{Float64}(undef, NITER, length(y))
ϕ = Matrix{Float64}(undef, NITER, length(y))
        
β₁ = Matrix{Float64}(undef, NITER, 2)
# η₀ = Vector{Float64}(undef, NITER)
# η₁ = Vector{Float64}(undef, NITER)
# ν = Vector{Float64}(undef, NITER)
τ₁ = Vector{Float64}(undef, NITER)

β₂ = Matrix{Float64}(undef, NITER, 2)
# ϕ₀ = Vector{Float64}(undef, NITER)
# ϕ₁ = Vector{Float64}(undef, NITER)
τ₂ = Vector{Float64}(undef, NITER)

ξ = Vector{Float64}(undef, NITER)

η[1,:] = log.(first.(shape.(fd)))
ϕ[1,:] = log.(first.(ExtendedExtremes.scale.(fd)))

β₁[1,1] = mean(η[1,:])
β₁[1,2] = 0.
τ₁[1] = std(η[1,:]) 

# ν[1] = mean(η[1,:])
# τ₁[1] = std(η[1,:]) 

β₂[1,1] = mean(ϕ[1,:])
β₂[1,2] = 0.
τ₂[1] = std(ϕ[1,:]) 

ξ[1] = 0.15

prior_xi = LocationScale(-.5, 1, Beta(9,6))

@showprogress for iter in 2:NITER
    
    for i in 1:m
       
        η̃ = η[iter-1,i] + randn()/5
        
        lr = loglike(y[i], n⁻[i], threshold, η̃, ϕ[iter-1,i], ξ[iter-1]) +
            logpdf(Normal(dot(X[i,:],β₁[iter-1,:]), τ₁[iter-1]), η̃) -
            loglike(y[i], n⁻[i], threshold, η[iter-1,i], ϕ[iter-1,i], ξ[iter-1]) - 
            logpdf(Normal(dot(X[i,:],β₁[iter-1,:]), τ₁[iter-1]), η[iter-1,i])

        if lr > log(rand())
            η[iter,i] = η̃
        else
            η[iter,i] = η[iter-1,i]
        end

        ϕ̃ = ϕ[iter-1,i] + randn()/5

        lr = loglike(y[i], n⁻[i], threshold, η[iter,i], ϕ̃, ξ[iter-1]) +
            logpdf(Normal(dot(X[i,:],β₂[iter-1,:]), τ₂[iter-1]), ϕ̃) -
            loglike(y[i], n⁻[i], threshold, η[iter,i], ϕ[iter-1,i], ξ[iter-1]) -
            logpdf(Normal(dot(X[i,:],β₂[iter-1,:]), τ₂[iter-1]), ϕ[iter-1,i])

        if lr > log(rand())
            ϕ[iter,i] = ϕ̃
        else
            ϕ[iter,i] = ϕ[iter-1,i]
        end  
        
    end
    
    v = X' * η[iter,:]
    avg = Q\v

    β₁[iter,:] = rand(MvNormal(avg, τ₁[1]^2 .* V))

    δ = η[iter,:] - X * β₁[iter,:]

    b = dot(δ, δ) + dot(β₁[iter,:], β₁[iter,:])

    τ₁[iter] = sqrt(rand(InverseGamma(2. + m/2, 1. + .5*b)))

    
    v = X' * ϕ[iter,:]
    avg = Q\v

    β₂[iter,:] = rand(MvNormal(avg, τ₂[iter-1]^2 .* V))

    δ = ϕ[iter,:] - X * β₂[iter,:]

    b = dot(δ, δ) + dot(β₂[iter,:], β₂[iter,:])

    τ₂[iter] = sqrt(rand(InverseGamma(2. + m/2, 1. + .5*b)))
    

    
    ξ̃ = ξ[iter-1] + randn()/100

    lr = sum(loglike(y[i], n⁻[i], threshold, η[iter,i], ϕ[iter,i], ξ̃) for i in 1:m) - 
        sum(loglike(y[i], n⁻[i], threshold, η[iter,i], ϕ[iter,i], ξ[iter-1]) for i in 1:m) +
        logpdf(prior_xi, ξ̃) - logpdf(prior_xi, ξ[iter-1])

    if lr > log(rand())
        ξ[iter] = ξ̃
    else
        ξ[iter] = ξ[iter-1]
    end
    
    
end

results = DataFrame(Iteration = 1:NITER, η₀ = β₁[:, 1], η₁ = β₁[:,2], τ₁=τ₁, ϕ₀=β₂[:,1], ϕ₁=β₂[:,2], τ₂ = τ₂, ξ = ξ)
CSV.write("power_results.csv", results)

df_eta = DataFrame(η, :auto)
CSV.write("power_eta.csv", df_eta)

df_phi = DataFrame(ϕ, :auto)
CSV.write("power_phi.csv", df_phi)
```
